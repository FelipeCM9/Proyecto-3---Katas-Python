Escribe un programa en Python que cree una lista de diccionarios con información de estudiantes (nombre, edad, calificación) y use filter para extraer a los estudiantes con una calificación mayor o igual a 90.

```python
estudiantes = [
    {"nombre": "Ana", "edad": 20, "calificacion": 8.5},
    {"nombre": "Luis", "edad": 22, "calificacion": 7.0},
    {"nombre": "María", "edad": 19, "calificacion": 9.2},
    {"nombre": "Carlos", "edad": 21, "calificacion": 6.5},
    {"nombre": "Lucía", "edad": 20, "calificacion": 9.8}
]

# for estudiante in estudiantes:
#     print(estudiante)

def calificacion(lista_estudiantes):
    def mayor_90(estudiante):
        if estudiante["calificacion"] > 9.0:
            return estudiante

    return list(filter(mayor_90, lista_estudiantes))

print(calificacion(estudiantes))
```
