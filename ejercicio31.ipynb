Enunciado:

Crea una función que solicite al usuario ingresar una lista de nombres y luego un nombre para buscar en esa lista. Si el nombre está en la lista, imprime un mensaje indicando que fue encontrado; de lo contrario, lanza una excepción.

```python
def buscar_nombre():
    nombres = []

    try:
        while True:
            nombre = input("Introduce un nombre (pulsa Enter para finalizar): ")

            if nombre == "":
                break

            nombres.append(nombre)

        nombre_buscar = input("¿Qué nombre quieres buscar?: ")

        if nombre_buscar in nombres:
            print("El nombre ha sido encontrado")
        else:
            raise ValueError("El nombre no está en la lista")

    except ValueError as error:
        print("Error:", error)

buscar_nombre()
```
