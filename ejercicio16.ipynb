Enunciado:

Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de todas las palabras que sean más largas que n. Usa la función filter().

```python

def cadena_texto(numero, cadenatexto):

    cadena_split = cadenatexto.split()

    def comprobar_palabra(cadena_split):
        if len(cadena_split) > numero:
            return cadena_split

    
    return list(filter(comprobar_palabra, cadena_split))

cadena_texto(3, "Felipe le encanta la buena musica")
```
