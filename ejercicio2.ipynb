Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

```python
numeros = [7, 23, 41, 56, 12, 89, 34, 65, 18, 92]

def multiplicar_numeros(numero):
    return numero * 2

resultado = map(multiplicar_numeros, numeros)

print(list(resultado))

```
