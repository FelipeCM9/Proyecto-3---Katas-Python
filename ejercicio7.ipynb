Enunciado:

Genera una función que convierta una lista de tuplas a una lista de strings. Usa la función map().

```python
personas = [
        ("Ana", 25),
        ("Carlos", 30),
        ("Lucía", 22),
        ("Miguel", 28),
        ("Sofía", 35),
        ("Felipe", 24),
        ("Yaneth", 56)
    ]

def mostrar_persona(persona):
        return persona[0] + " - " + str(persona[1])

resultado = list(map(mostrar_persona, personas))

print(resultado)
```
