Dada una lista numérica, obtén el producto total de los valores. Usa la función reduce().


```python
from functools import reduce

numeros = [2, 3, 4, 5, 6]

def multiplicar_numeros(lista_numeros):
    return reduce(lambda acumulado, numero_siguiente: acumulado * numero_siguiente, lista_numeros)

print(multiplicar_numeros(numeros))
```
