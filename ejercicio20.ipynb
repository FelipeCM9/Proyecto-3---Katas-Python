Para una lista con elementos de tipo integer y string, obtén una nueva lista solo con los valores int. Usa la función filter().

```python
lista = [1, "hola", 25, "casa", 8, "perro", 42, "azul", 3, "mesa"]

def only_int(numero):
    if isinstance(numero, int):
        return True
    else:
        return False

resultado = list(filter(only_int, lista))

print(resultado)
```
