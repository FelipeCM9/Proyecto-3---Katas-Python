Concatena una lista de palabras. Usa la función reduce().

```python
from functools import reduce

palabras = ["Lionel", "Messi", "es", "uno", "de", "los", "mejores", "futbolistas", "del", "mundo"]

def unir_palabras(lista_palabras):
    return reduce(lambda primera_palabra, siguiente_palabra: f"{primera_palabra} {siguiente_palabra}", lista_palabras)

print(unir_palabras(palabras))




    
```
