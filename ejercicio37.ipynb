Enunciado:

Genera un programa que nos indique si es de noche, de día o de tarde según la hora proporcionada por el usuario.

```python
def hora_dia():

    hora = int(input("Indique una hora (0-23): "))

    if hora < 0 or hora > 23:
        return "La hora no es valida, ingresa una hora entre las (0 y 23) horas"
    elif hora >= 0 and hora <= 6:
        return "Es de madrugada"
    elif hora > 6 and hora <= 12:
        return "Es de mañana"
    elif hora > 12 and hora <= 20:
        return "Es de tarde"
    else:
        return "Es de noche"

print(hora_dia())
```
