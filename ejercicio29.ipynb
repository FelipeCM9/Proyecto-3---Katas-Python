Enunciado:

Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#' excepto los últimos cuatro.

```python
numero = 3424234243

def ultimos_numeros(numero):
    #Parseamos el numero de int a string
    numero_string = str(numero)
    
    #Obtenemos los ultimos 4 digitos de este numero
    ultimos_cuatro = numero_string[-4:]
    # return ultimos_cuatro

    numeros_enmascarados = max(0, len(numero_string) - 4)
    resultado = "#" * numeros_enmascarados + ultimos_cuatro
    return resultado

print(ultimos_numeros(numero))

```
