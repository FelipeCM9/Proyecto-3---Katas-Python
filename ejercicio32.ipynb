Enunciado:

Crea una función que tome un nombre completo y una lista de empleados, busque el nombre en la lista y devuelva el puesto del empleado si se encuentra; de lo contrario, devuelve un mensaje indicando que la persona no trabaja aquí.

```python
empleados = [
    {"nombre": "Ana Garcia", "puesto": "Recepcionista"},
    {"nombre": "Luis Martinez", "puesto": "Programador"},
    {"nombre": "Maria Sanchez", "puesto": "Contable"},
    {"nombre": "Carlos Fernandez", "puesto": "Diseñador gráfico"},
    {"nombre": "Laura Gomez", "puesto": "Directora"},
    {"nombre": "Pedro Rodríguez", "puesto": "Técnico informático"}
]

# for empleado in empleados:
#     print(empleado["nombre"])

def busca_nombre(nombre_buscado, lista_empleados):

    for empleado in lista_empleados:
        if empleado["nombre"] == nombre_buscado:
            return f"El puesto de trabajo de {empleado["nombre"]} es {empleado["puesto"]}"
    
    return "Esta persona no trabaja aquí"


print(busca_nombre("Luis Martinez", empleados))
print(busca_nombre("Ana Garcia", empleados))
print(busca_nombre("Carlos Fernandez", empleados))
print(busca_nombre("Felipe", empleados))

```
