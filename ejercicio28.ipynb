Enunciado:

Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

```python
numeros = [3, 8, 5, 2, 8, 3, 9]

def num_duplicado(lista_numeros):
    vistos = []
    
    for num in lista_numeros:
        if num not in vistos:
            vistos.append(num)
        else:
            return f"El primer numero duplicado es {num}"

print(num_duplicado(numeros))
```
