Crea una función llamada procesar_texto
Procesa un texto según la opción especificada: contar_palabras, reemplazar_palabras o eliminar_palabra.
Código a seguir:
Crear una función contar_palabras que cuente el número de veces que aparece cada palabra en el texto y devuelva un diccionario.
Crear una función reemplazar_palabras para sustituir una palabra_original por una palabra_nueva en el texto y devolver el texto modificado.
Crear una función eliminar_palabra que elimine una palabra del texto y devuelva el texto sin ella.
Crear la función procesar_texto que reciba un texto, una opción ("contar", "reemplazar", "eliminar") y un número variable de argumentos según la opción elegida.
Caso de uso:
Verificar el funcionamiento completo de procesar_texto.

```python
# Frase a utilizar en los ejemplos
frase = "El mundo es magico, tan magico es el mundo que tiene mil colores diferentes cada dia"
# Con el .split separo la frase por palabras y la añado a una lista
palabras = frase.split()


# Metodo para contar la palabra seleccionada y poder ver cuantas veces esta en la frase
def contar_palabras(palabras, palabra):
    cont = 0
    for i in palabras:
        if i == palabra:
            cont += 1
    return f"La palabra {palabra} se visualiza en la frase {cont} veces"


# Metodo para reemplazar palabras de la frase
def reemplazar_palabra(frase, palabra_original, palabra_nueva):
    if palabra_original not in frase:
        return "La palabra no se visualiza en la frase"
    else:
        texto_modificado = frase.replace(palabra_original, palabra_nueva)
        return texto_modificado
                

# Metodo para eliminar palabras de la frase
def eliminar_palabra(palabras, palabra_eliminar):
    if palabra_eliminar not in palabras:
        return "No es posible eliminar una palabra que no se visualiza la frase"
    else:
        nuevas_palabras = []

        for palabra in palabras:
            if palabra != palabra_eliminar:
                nuevas_palabras.append(palabra)
        return nuevas_palabras


# Metodo para hacer la prueba de los metodos anteriormente realizados
def procesar_texto(texto):

    if texto == "contar":
        return contar_palabras(palabras, "mundo")

    if texto == "reemplazar":
        return reemplazar_palabra(frase, "mundo", "pipe")

    if texto == "eliminar":
        return eliminar_palabra(palabras, "mundo")

    
print(procesar_texto("eliminar"))

```
