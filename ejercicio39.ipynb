Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

```python
from math import pi

def calcular_area(figura, tupla):

    if  figura == "rectangulo":
            base, altura = tupla
            area = base * altura
            return area
    elif figura == "circulo":
            radio = tupla
            area = pi * (radio ** 2)
            return area
    elif figura == "triangulo":
           base, altura = tupla
           area = base * altura / 2
           return area
    else:
           return "Figura no valida"


print(calcular_area("rectangulo", (4,2)))
print(calcular_area("circulo", (4)))
print(calcular_area("triangulo", (8,4)))
```
