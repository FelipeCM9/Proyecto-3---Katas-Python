Calcula la diferencia total en los valores de una lista. Usa la función reduce().

```python
from functools import reduce

numeros = [50, 10, 5, 2, 4, 7]

def restar_numeros(lista_numeros):
    return reduce(lambda acumulado, numero_siguiente: acumulado - numero_siguiente, lista_numeros)

print(restar_numeros(numeros))
```
