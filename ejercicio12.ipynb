Enunciado:

Genera una función que, al recibir una frase, devuelva una lista con la longitud de cada palabra. Usa la función map().

```python
def frases(frase):
    nueva_frase = frase.split() # separa cada palabra de la frase
    longitud_frase = map(len, nueva_frase) # sacamos la longitud de cada palabra de la frase
    return list(longitud_frase) #la convertimos en una lista

frases("Me llamo Felipe y soy un loco")
```
