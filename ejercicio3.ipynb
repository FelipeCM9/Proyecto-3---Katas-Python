Enunciado:

Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

```python
lista = ["balón", "baloncesto", "fútbol", "balonmano", "jugador", "balear", "portería"]

def listapalabras (lista, palabraobjetivo):
    
    #lista donde guardo las palabras que contienen la palabra objetivo
    resultado = []

    #for para recorrer la lista
    for palabra in lista:
        if palabraobjetivo in palabra:
            resultado.append(palabra)

    print(resultado)

listapalabras(lista, "bal")

```
