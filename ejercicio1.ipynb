Enunciado:

Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def frecuencia_letras(texto):
    frecuencias = {}

    for letra in texto:
        if letra == " ":
            continue
        
        if letra in frecuencias:
            #si la letra ya esta en el diccionario frecuencias sumas
            frecuencias[letra] += 1
        else:
            #si la letra no esta aun en el diccionario la agregas con el valor de 1 como inicial
            frecuencias[letra] = 1

    return frecuencias


frecuencia_letras("hola mundo")
```
