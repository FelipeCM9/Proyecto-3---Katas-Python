Enunciado:

Crea una función que calcule el cubo de un número dado mediante una función lambda.


```python
# ** 3
# Esto multiplica el numero que le estoy pasando a la funcion tres veces, en este caso 7x7x7

def calcular_cubo(numero_argumento):
    funcion_lambda = lambda numero_argumento: numero_argumento ** 3
    
    #Aqui estoy devolviendo la funcion lambda para que se ejecute con el numero que le voy a pasar por argumento
    return funcion_lambda(numero_argumento)

print(calcular_cubo(7))
```
