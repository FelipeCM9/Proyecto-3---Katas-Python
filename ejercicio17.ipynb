Enunciado:

Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2] corresponde al número 572. Usa la función reduce().

```python
from functools import reduce

numeros = [2, 3, 4, 5]

def convertir_numeros(lista_numeros):
    return reduce(lambda acumulado, numero_siguiente: acumulado * 10 + numero_siguiente, lista_numeros)

print(convertir_numeros(numeros))



```
