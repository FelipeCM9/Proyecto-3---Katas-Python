Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map().

```python
lista1 = [25, 40, 15, 60, 80, 35, 50, 90, 10, 70]

lista2 = [10, 15, 5, 20, 30, 10, 25, 40, 5, 30]

def restar (a, b):
    return a - b

def resta_listas (lista1, lista2):
    resultado = list(map(restar, lista1, lista2))
    return resultado

resultado = resta_listas(lista1, lista2)

print("Lista 1:", lista1)
print("Lista 2:", lista2)
print("---------------------------------------------------")
print("Diferencia:", resultado)

```
