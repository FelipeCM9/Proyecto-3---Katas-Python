Enunciado:

Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().

```python
palabras = [
    "casa",
    "coche",
    "perro",
    "gato",
    "camino",
    "mesa",
    "silla",
    "computadora",
    "cielo",
    "sol",
    "libro",
    "lapiz",
    "telefono",
    "puerta",
    "ventana",
    "montaña", 
    "mar",
    "rio",
    "flor",
    "arbol"
]

def filtrar_palabras(letra , lista_palabras):

    def comprobar_palabra(palabra):
        if palabra[0] == letra:
            return palabra

    return list(filter(comprobar_palabra, lista_palabras))

filtrar_palabras("c", palabras)
    
```
