Enunciado:

Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
def anagramas(palabra1, palabra2):
    palabra1_sorted = sorted(palabra1)
    palabra2_sorted = sorted(palabra2)

    print(palabra1_sorted)
    print(palabra2_sorted)

    if len(palabra1) == len(palabra2):
        if palabra1_sorted == palabra2_sorted:
            return f"La palabra {palabra1} y la palabra {palabra2} son anamagras"
        else:
            return "Aunque las palabras tienen la misma longitud, no son palabras anagramas"
    else:
        return "Las palabras no tienen la misma longitud"
    

print(anagramas("amor", "pintor"))
```
