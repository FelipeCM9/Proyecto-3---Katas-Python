Enunciado:

Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. Usa la función filter().

```python
mascotas = [
    "Perro",
    "Gato",
    "Hurón",
    "Conejo",
    "Hámster",
    "Cobaya",
    "Canario",
    "Periquito",
    "Pez",
    "Tortuga",
    "Mapache",
    "Tigre",
    "Serpiente Pitón",
    "Cocodrilo",
    "Oso"
]

def filtrar_mascotas(lista_mascotas):

    mascotas_prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]

    def mascota_permitida(mascota):
        return mascota not in mascotas_prohibidas  ### si la mascota no esta dentro de la lista de mascotas prohibidas

    return list(filter(mascota_permitida, lista_mascotas))

resultado = filtrar_mascotas(mascotas)
print(resultado)

```
