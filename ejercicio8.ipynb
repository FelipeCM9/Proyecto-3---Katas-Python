Enunciado:

Escribe un programa que pida al usuario dos números e intente dividirlos. Si el usuario ingresa un valor no numérico o intenta dividir por cero, maneja esas excepciones de manera adecuada y muestra un mensaje indicando si la división fue exitosa o no.

```python
num1 = input("Escribe el primer número: ")
num2 = input("Escribe el segundo número: ")

def division(num1, num2):
    try:
        num1 = float(num1)
        num2 = float(num2)

        div = num1 / num2

        print("División exitosa")
        return "Resultado:", div

    except ValueError:
        return "Valor ingresado no es numérico"

    except ZeroDivisionError:
        return "Estás intentando dividir por cero"

print(division(num1, num2))
```
