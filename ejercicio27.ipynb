Enunciado:

Crea una función que calcule el promedio de una lista de números.

```python
numeros = [8, 12, 15, 20, 25, 10, 5, 7, 22, 26, 9]

def promedio(lista_numeros):
    suma_lista = sum(lista_numeros)
    promedio = suma_lista / len(lista_numeros)
    return promedio

print(promedio(numeros))

```
